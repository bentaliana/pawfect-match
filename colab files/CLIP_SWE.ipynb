```python
import torch
import clip
from PIL import Image
import itertools

device = "cuda" if torch.cuda.is_available() else "cpu"
model, preprocess = clip.load("ViT-B/32", device=device)

image_path = 'drive/My Drive/Project/Project data/dog.jpg.webp'  # Replace with your image path
image = preprocess(Image.open(image_path)).unsqueeze(0).to(device)

adjectives = ["small", "fluffy", "sleek", "spotted", "striped", "soft", "shiny", "bright-eyed", "whiskered", "short-haired", "long-haired"]
animals = ["dog", "cat"]


combinations = list(itertools.product(adjectives, animals))
sentences = [f"{adj} {animal}" for adj, animal in combinations]

text = clip.tokenize(sentences).to(device)

with torch.no_grad():
    image_features = model.encode_image(image)
    text_features = model.encode_text(text)

similarity = (image_features @ text_features.T).softmax(dim=-1)

best_match_idx = similarity.argmax().item()
best_description = sentences[best_match_idx]

print(f"Best description: {best_description}")
```

```python
import torch
import clip
from PIL import Image

device = "cuda" if torch.cuda.is_available() else "cpu"
model, preprocess = clip.load("ViT-B/32", device=device)

image_path = 'Zrzut ekranu 2024-11-14 o 13.50.07.png'
image = preprocess(Image.open(image_path)).unsqueeze(0).to(device)

adjectives = ["spotted", "blue-eyed", "fluffy", "long-haired", "sleek", "soft", "whiskered", "bright-eyed"]
animals = ["dog", "cat"]

animal_sentences = animals
animal_text = clip.tokenize(animal_sentences).to(device)

with torch.no_grad():
    image_features = model.encode_image(image)
    animal_features = model.encode_text(animal_text)
    animal_similarity = (image_features @ animal_features.T).softmax(dim=-1).squeeze(0)

best_animal_idx = animal_similarity.argmax().item()
identified_animal = animals[best_animal_idx]

sentences = [f"{adj} {identified_animal}" for adj in adjectives]
text = clip.tokenize(sentences).to(device)

with torch.no_grad():
    text_features = model.encode_text(text)
    similarity = (image_features @ text_features.T).softmax(dim=-1).squeeze(0)

threshold = 0.05
suitable_descriptions = [sentences[i].split()[0] for i in range(len(similarity)) if similarity[i] > threshold]

unique_adjectives = sorted(set(suitable_descriptions))

result = unique_adjectives + [identified_animal]

print(result)


```

```python
import unittest
import torch
import clip
from PIL import Image

class TestClipImageDescription(unittest.TestCase):

    @classmethod
    def setUpClass(cls):
        cls.device = "cuda" if torch.cuda.is_available() else "cpu"
        cls.model, cls.preprocess = clip.load("ViT-B/32", device=cls.device)

    def setUp(self):

        self.image_path = 'Zrzut ekranu 2024-11-14 o 13.50.07.png'
        self.image = self.preprocess(Image.open(self.image_path)).unsqueeze(0).to(self.device)


        self.adjectives = ["spotted", "blue-eyed", "fluffy", "long-haired", "sleek", "soft", "whiskered", "bright-eyed"]
        self.animals = ["dog", "cat"]

    def test_identify_animal(self):

        animal_text = clip.tokenize(self.animals).to(self.device)

        with torch.no_grad():
            image_features = self.model.encode_image(self.image)
            animal_features = self.model.encode_text(animal_text)
            animal_similarity = (image_features @ animal_features.T).softmax(dim=-1).squeeze(0)


        best_animal_idx = animal_similarity.argmax().item()
        identified_animal = self.animals[best_animal_idx]

        self.assertAlmostEqual(animal_similarity.sum().item(), 1.0, places=5, msg="Similarity scores do not sum to 1")
        self.assertIn(identified_animal, self.animals, "Identified animal is not in the predefined list")

    def test_generate_descriptions(self):

        identified_animal = "dog"
        sentences = [f"{adj} {identified_animal}" for adj in self.adjectives]
        text = clip.tokenize(sentences).to(self.device)

        with torch.no_grad():
            image_features = self.model.encode_image(self.image)
            text_features = self.model.encode_text(text)
            similarity = (image_features @ text_features.T).softmax(dim=-1).squeeze(0)

        threshold = 0.05
        suitable_descriptions = [sentences[i].split()[0] for i in range(len(similarity)) if similarity[i] > threshold]

        unique_adjectives = sorted(set(suitable_descriptions))


        self.assertTrue(len(unique_adjectives) > 0, "No adjectives matched the similarity threshold")

        result_descriptions = [f"{adj} {identified_animal}" for adj in unique_adjectives]
        for desc in result_descriptions:
            self.assertIn(identified_animal, desc, f"Description '{desc}' should include the identified animal '{identified_animal}'")

    def test_output_format(self):
        identified_animal = "dog"
        result = ["fluffy", "soft", "whiskered", identified_animal]

        self.assertIsInstance(result, list, "Result is not a list")
        self.assertEqual(result[-1], identified_animal, "Identified animal should be the last item in the list")
        self.assertTrue(all(isinstance(item, str) for item in result), "All items in the result should be strings")

# Run the tests
unittest.main(argv=[''], exit=False)
```
